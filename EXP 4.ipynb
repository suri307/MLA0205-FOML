```python
import numpy as np

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

np.random.seed(42)

W1 = np.random.uniform(-1, 1, (2, 4))
b1 = np.zeros((1, 4))

W2 = np.random.uniform(-1, 1, (4, 1))
b2 = np.zeros((1, 1))

learning_rate = 0.5
epochs = 20000

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

for epoch in range(epochs):
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    output_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(output_input)

    output_error = predicted_output - y
    output_delta = output_error * predicted_output * (1 - predicted_output)

    hidden_error = np.dot(output_delta, W2.T)
    hidden_delta = hidden_error * hidden_output * (1 - hidden_output)

    W2 -= learning_rate * np.dot(hidden_output.T, output_delta) / len(X)
    b2 -= learning_rate * np.mean(output_delta, axis=0, keepdims=True)

    W1 -= learning_rate * np.dot(X.T, hidden_delta) / len(X)
    b1 -= learning_rate * np.mean(hidden_delta, axis=0, keepdims=True)

hidden_output = sigmoid(np.dot(X, W1) + b1)
predicted_output = sigmoid(np.dot(hidden_output, W2) + b2)

print("Input Data:")
print(X)

print("\nActual Output:")
print(y.flatten())

print("\nPredicted Output:")
print(np.round(predicted_output.flatten(), 4))

print("\nFinal Classification:")
print((predicted_output >= 0.5).astype(int).flatten())
```
