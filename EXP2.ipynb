```python
import csv

with open("/content/enjoysport.csv", "r") as file:
    data = list(csv.reader(file))

header = data[0]
training_data = data[1:]

num_attributes = len(header) - 1

S = ["0"] * num_attributes
G = [["?"] * num_attributes]

for index, row in enumerate(training_data, start=1):
    attributes = row[:-1]
    target = row[-1].lower()

    print("\nTraining Example", index, ":", row)

    if target == "yes":
        G = [g for g in G if all(
            g[i] == "?" or g[i] == attributes[i]
            for i in range(num_attributes)
        )]

        for i in range(num_attributes):
            if S[i] == "0":
                S[i] = attributes[i]
            elif S[i] != attributes[i]:
                S[i] = "?"

    else:
        new_G = []

        for g in G:
            if all(g[i] == "?" or g[i] == attributes[i]
                   for i in range(num_attributes)):
                for i in range(num_attributes):
                    if g[i] == "?":
                        for value in set(r[i] for r in training_data):
                            if value != attributes[i] and value != "":
                                new_hypothesis = g.copy()
                                new_hypothesis[i] = value

                                if all(
                                    new_hypothesis[j] == "?" or
                                    new_hypothesis[j] == S[j] or
                                    S[j] == "0"
                                    for j in range(num_attributes)
                                ):
                                    new_G.append(new_hypothesis)
            else:
                new_G.append(g)

        G = new_G

    print("Specific Boundary (S):", S)
    print("General Boundary (G):", G)

print("\nFinal Specific Boundary:")
print(S)

print("\nFinal General Boundary:")
for hypothesis in G:
    print(hypothesis)
```
