```python
import csv

with open("/content/enjoysport.csv", "r") as file:
    data = list(csv.reader(file))

header = data[0]
training_data = data[1:]

hypothesis = ["0"] * (len(header) - 1)

print("Initial Hypothesis:")
print(hypothesis)

for index, row in enumerate(training_data, start=1):
    print("\nTraining Example", index, ":")
    print(row)

    if row[-1].lower() == "yes":
        print("Positive Example")

        for i in range(len(hypothesis)):
            if hypothesis[i] == "0":
                hypothesis[i] = row[i]
            elif hypothesis[i] != row[i]:
                hypothesis[i] = "?"

        print("Updated Hypothesis:")
    else:
        print("Negative Example")
        print("Hypothesis remains unchanged:")

    print(hypothesis)

print("\nFinal Most Specific Hypothesis:")
print(hypothesis)
```
